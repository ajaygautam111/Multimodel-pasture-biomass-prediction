```python
# ============================================================
# JULIAN COSINE ENCODING
# ============================================================

def julian_cos_encoding_from_date(
    df: pd.DataFrame,
    date_col: str,
    cos_JD: str = "Julian_Cos"
) -> pd.DataFrame:
    """
    Convert calendar date into a cyclic cosine representation.

    Steps:
    1. Convert date column to datetime.
    2. Extract year and day-of-year.
    3. Determine whether each year has 365 or 366 days.
    4. Convert day-of-year to angular position.
    5. Apply cosine transformation.
    """

    df = df.copy()

    df[date_col] = pd.to_datetime(
        df[date_col],
        errors="coerce"
    )

    years = df[date_col].dt.year
    julian_days = df[date_col].dt.dayofyear

    # Account for leap years
    days_in_year = np.where(
        ((years % 4 == 0) & (years % 100 != 0))
        | (years % 400 == 0),
        366,
        365
    )

    radians = (
        2
        * np.pi
        * julian_days
        / days_in_year
    )

    df[cos_JD] = np.cos(radians)

    return df
```

```python
# ============================================================
# SEASON FROM JULIAN_COS
# ============================================================

def season_from_julian_cos(jcos, date_val):

    if pd.isna(jcos) or pd.isna(date_val):
        return np.nan

    # Ensure jcos lies within valid arccos limits
    jcos = float(
        np.clip(
            jcos,
            -1.0,
            1.0
        )
    )

    # Determine year length
    year = date_val.year

    is_leap = (
        (year % 4 == 0 and year % 100 != 0)
        or
        (year % 400 == 0)
    )

    days_in_year = 366 if is_leap else 365

    # First possible angle
    theta1 = np.arccos(jcos)

    # Second possible angle because cosine is symmetric
    theta2 = (2 * np.pi) - theta1

    # Recover the two possible DOYs
    doy1 = int(
        round(
            theta1
            / (2 * np.pi)
            * days_in_year
        )
    )

    doy2 = int(
        round(
            theta2
            / (2 * np.pi)
            * days_in_year
        )
    )

    # Keep within valid DOY range
    doy1 = max(1, min(days_in_year, doy1))
    doy2 = max(1, min(days_in_year, doy2))

    # Approximate middle DOY of each month
    month_mid_doy = {
        1: 15,
        2: 46,
        3: 74,
        4: 105,
        5: 135,
        6: 166,
        7: 196,
        8: 227,
        9: 258,
        10: 288,
        11: 319,
        12: 349
    }

    target = month_mid_doy.get(
        date_val.month,
        180
    )

    # Select the candidate DOY closest to the actual month
    doy = (
        doy1
        if abs(doy1 - target) <= abs(doy2 - target)
        else doy2
    )

    # Meteorological/DOY season classification
    if 79 <= doy <= 172:
        return "Spring"

    elif 173 <= doy <= 265:
        return "Summer"

    elif 266 <= doy <= 354:
        return "Fall"

    else:
        return "Winter"
```

```python
df["Season"] = df.apply(
    lambda row: season_from_julian_cos(
        row["Julian_Cos"],
        row["Date"]
    ),
    axis=1
)

df[["Date", "Julian_Cos", "Season"]].head(20)
```

```python
# ============================================================
# LOAD WEATHER DATA
# ============================================================

dz = pd.read_csv(weather_file)

print("Weather data shape:", dz.shape)

print("\nWeather columns:")
print(dz.columns.tolist())

dz.head()
```

```python
# ============================================================
# COLUMN DEFINITIONS
# ============================================================

FARM_DATE_COL = "Date"
WEATHER_DATE_COL = "date"

PREC_COL = "precipitation_in"
GDD_COL = "GDD"
ET_COL = "ETo_in"

SWB_PCT_COL = "Available_Soil_Water_%"
SWB_IN_COL = "Available_Soil_Water_in"

TMIN_COL = "temperature_min_F"
TMAX_COL = "temperature_max_F"
TAVG_COL = "temperature_avg_F"
```

```python
# ============================================================
# PREPARE DATA
# ============================================================

da = df.copy()
w = dz.copy()

da[FARM_DATE_COL] = pd.to_datetime(
    da[FARM_DATE_COL],
    errors="coerce"
)

w[WEATHER_DATE_COL] = pd.to_datetime(
    w[WEATHER_DATE_COL],
    errors="coerce"
)

print(
    "Invalid main-data dates:",
    da[FARM_DATE_COL].isna().sum()
)

print(
    "Invalid weather dates:",
    w[WEATHER_DATE_COL].isna().sum()
)
```

```python
# ============================================================
# CREATE DAILY AVERAGE TEMPERATURE
# ============================================================

if TAVG_COL not in w.columns:

    if (
        TMIN_COL not in w.columns
        or TMAX_COL not in w.columns
    ):
        raise ValueError(
            "Weather dataset is missing "
            "temperature_min_F and/or temperature_max_F."
        )

    w[TAVG_COL] = (
        w[TMIN_COL]
        + w[TMAX_COL]
    ) / 2

print(
    w[
        [
            WEATHER_DATE_COL,
            TMIN_COL,
            TMAX_COL,
            TAVG_COL
        ]
    ].head()
)
```

```python
# ============================================================
# REQUIRED COLUMN CHECK
# ============================================================

needed_weather = [
    WEATHER_DATE_COL,
    PREC_COL,
    GDD_COL,
    ET_COL,
    SWB_PCT_COL,
    SWB_IN_COL,
    TAVG_COL,
    TMIN_COL,
    TMAX_COL
]

missing_weather_cols = [
    col
    for col in needed_weather
    if col not in w.columns
]

if missing_weather_cols:
    raise ValueError(
        "Weather dataset is missing these columns:\n"
        + "\n".join(missing_weather_cols)
    )

if FARM_DATE_COL not in da.columns:
    raise ValueError(
        f"Main dataset is missing: {FARM_DATE_COL}"
    )

print("All required columns found.")
```

```python
# ============================================================
# SORT WEATHER DATA
# ============================================================

w = (
    w
    .sort_values(WEATHER_DATE_COL)
    .set_index(WEATHER_DATE_COL)
)

w.head()
```

```python
# ============================================================
# FILL FLOW VARIABLES
# ============================================================

w[PREC_COL] = w[PREC_COL].fillna(0)
w[GDD_COL] = w[GDD_COL].fillna(0)
w[ET_COL] = w[ET_COL].fillna(0)
```

```python
# ============================================================
# ROLLING WEATHER FEATURES
# ============================================================

def window_feats(window: str):

    # --------------------------------------------------------
    # SUM variables
    # --------------------------------------------------------

    sums = (
        w[
            [
                PREC_COL,
                GDD_COL,
                ET_COL,
                TAVG_COL,
                TMIN_COL,
                TMAX_COL
            ]
        ]
        .rolling(
            window,
            closed="left"
        )
        .sum()
        .add_suffix(
            f"_{window.lower()}_sum"
        )
    )

    # --------------------------------------------------------
    # MEAN variables
    # --------------------------------------------------------

    avgs = (
        w[
            [
                SWB_PCT_COL,
                SWB_IN_COL,
                TAVG_COL,
                TMIN_COL,
                TMAX_COL
            ]
        ]
        .rolling(
            window,
            closed="left"
        )
        .mean()
        .add_suffix(
            f"_{window.lower()}_avg"
        )
    )

    return pd.concat(
        [sums, avgs],
        axis=1
    )
```

```python
# ============================================================
# CREATE WEATHER WINDOWS
# ============================================================

feat_7d = window_feats("7D")
feat_14d = window_feats("14D")
feat_21d = window_feats("21D")

feat = pd.concat(
    [
        feat_7d,
        feat_14d,
        feat_21d
    ],
    axis=1
)

feat = feat.reset_index()

print("Weather feature shape:", feat.shape)

feat.head()
```

```python
# ============================================================
# COMPLETE WORKFLOW
# 1. Load data
# 2. Remove erroneous height values <= 0
# 3. Apply Huber outlier filtering
# 4. Separate clean observations and outliers
# 5. Calculate regression statistics
# 6. Create publication-style before/after figure
# 7. Export JPEG and filtered datasets
# ============================================================


# ============================================================
# CELL 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import HuberRegressor, LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error


# ============================================================
# CELL 2. PLOTTING DEFAULTS
# ============================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.labelsize": 12,
    "axes.titlesize": 12,
    "axes.titleweight": "bold",
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 9,
    "figure.dpi": 350
})


# ============================================================
# CELL 3. LOAD DATA
# ============================================================

file_path = (
    r"C:\Users\ag2c2\OneDrive - University of Missouri"
    r"\RESEARCH\Testbed Paper\data\TB_24_25_all_data.csv"
)

df = pd.read_csv(file_path)

print("Dataset dimensions:", df.shape)

display(df.head())


# ============================================================
# CELL 4. DEFINE VARIABLES
# ============================================================

HEIGHT_COL = "PT_Height(mm)"
BIOMASS_COL = "Total Biomass (kg/ha)"

# Huber threshold requested
HUBER_EPSILON = 2.57


# ============================================================
# CELL 5. CONVERT VARIABLES TO NUMERIC
# ============================================================

df[HEIGHT_COL] = pd.to_numeric(
    df[HEIGHT_COL],
    errors="coerce"
)

df[BIOMASS_COL] = pd.to_numeric(
    df[BIOMASS_COL],
    errors="coerce"
)


# ============================================================
# CELL 6. INITIAL DATA QUALITY CHECK
# ============================================================

n_original = len(df)

n_height_le_zero = (
    df[HEIGHT_COL] <= 0
).sum()

n_height_missing = (
    df[HEIGHT_COL].isna()
).sum()

n_biomass_missing = (
    df[BIOMASS_COL].isna()
).sum()


print("=" * 60)
print("INITIAL DATA QUALITY CHECK")
print("=" * 60)

print(
    f"Original observations        : {n_original}"
)

print(
    f"Height <= 0                  : {n_height_le_zero}"
)

print(
    f"Missing/non-numeric height   : {n_height_missing}"
)

print(
    f"Missing/non-numeric biomass  : {n_biomass_missing}"
)


# ============================================================
# CELL 7. EXTRACT ERRONEOUS HEIGHT RECORDS
# ============================================================

erroneous_height_df = df.loc[
    df[HEIGHT_COL].notna()
    & (df[HEIGHT_COL] <= 0)
].copy()


print(
    f"\nNumber of erroneous height records: "
    f"{len(erroneous_height_df)}"
)

display(erroneous_height_df)


# ============================================================
# CELL 8. REMOVE ERRONEOUS HEIGHT VALUES
# ============================================================

df_height_clean = df.loc[
    df[HEIGHT_COL].notna()
    & df[BIOMASS_COL].notna()
    & (df[HEIGHT_COL] > 0)
].copy()


print("\n" + "=" * 60)
print("AFTER ERRONEOUS HEIGHT FILTERING")
print("=" * 60)

print(
    f"Original records                : {len(df)}"
)

print(
    f"Height <= 0 removed             : {n_height_le_zero}"
)

print(
    f"Missing height excluded         : {n_height_missing}"
)

print(
    f"Missing biomass excluded        : {n_biomass_missing}"
)

print(
    f"Records entering Huber analysis : {len(df_height_clean)}"
)


# ============================================================
# CELL 9. FIT HUBER ROBUST REGRESSION
# ============================================================

X = df_height_clean[
    [HEIGHT_COL]
].values

y = df_height_clean[
    BIOMASS_COL
].values


huber = HuberRegressor(
    epsilon=HUBER_EPSILON,
    alpha=0.0,
    max_iter=1000
)

huber.fit(
    X,
    y
)


print("\n" + "=" * 60)
print("HUBER ROBUST REGRESSION")
print("=" * 60)

print(
    f"Huber threshold, epsilon : {HUBER_EPSILON}"
)

print(
    f"Huber slope              : {huber.coef_[0]:.4f}"
)

print(
    f"Huber intercept          : {huber.intercept_:.4f}"
)

print(
    f"Huber scale              : {huber.scale_:.4f}"
)

print(
    f"Detected outliers        : {huber.outliers_.sum()}"
)


# ============================================================
# CELL 10. ADD HUBER RESULTS TO DATAFRAME
# ============================================================

df_huber = df_height_clean.copy()

df_huber["Huber_Outlier"] = (
    huber.outliers_
)

df_huber["Huber_Predicted_Biomass"] = (
    huber.predict(X)
)

df_huber["Huber_Residual"] = (
    df_huber[BIOMASS_COL]
    - df_huber["Huber_Predicted_Biomass"]
)

df_huber["Huber_Scaled_Residual"] = (
    df_huber["Huber_Residual"]
    / huber.scale_
)


display(
    df_huber[
        [
            HEIGHT_COL,
            BIOMASS_COL,
            "Huber_Predicted_Biomass",
            "Huber_Residual",
            "Huber_Scaled_Residual",
            "Huber_Outlier"
        ]
    ].head()
)


# ============================================================
# CELL 11. SEPARATE OUTLIERS AND CLEAN DATA
# ============================================================

df_outliers = df_huber.loc[
    df_huber["Huber_Outlier"]
].copy()

df_clean = df_huber.loc[
    ~df_huber["Huber_Outlier"]
].copy()


n_huber_outliers = len(
    df_outliers
)

n_clean = len(
    df_clean
)


print("\n" + "=" * 60)
print("FINAL FILTERING SUMMARY")
print("=" * 60)

print(
    f"Erroneous heights removed : {n_height_le_zero}"
)

print(
    f"Huber outliers removed    : {n_huber_outliers}"
)

print(
    f"Final clean observations  : {n_clean}"
)


# ============================================================
# CELL 12. VIEW DETECTED OUTLIERS
# ============================================================

outlier_columns = [
    HEIGHT_COL,
    BIOMASS_COL,
    "Huber_Predicted_Biomass",
    "Huber_Residual",
    "Huber_Scaled_Residual",
    "Huber_Outlier"
]


display(
    df_outliers[
        outlier_columns
    ].sort_values(
        "Huber_Scaled_Residual",
        key=np.abs,
        ascending=False
    )
)


# ============================================================
# CELL 13. REGRESSION STATISTICS FUNCTION
# ============================================================

def regression_statistics(
    data,
    x_col,
    y_col
):

    X_reg = data[
        [x_col]
    ].values

    y_reg = data[
        y_col
    ].values


    model = LinearRegression()

    model.fit(
        X_reg,
        y_reg
    )


    y_pred = model.predict(
        X_reg
    )


    return {
        "model": model,

        "slope": model.coef_[0],

        "intercept": model.intercept_,

        "r2": r2_score(
            y_reg,
            y_pred
        ),

        "rmse": np.sqrt(
            mean_squared_error(
                y_reg,
                y_pred
            )
        ),

        "mae": mean_absolute_error(
            y_reg,
            y_pred
        ),

        "n": len(data)
    }


# ============================================================
# CELL 14. CALCULATE BEFORE/AFTER STATISTICS
# ============================================================

stats_before = regression_statistics(
    df_height_clean,
    HEIGHT_COL,
    BIOMASS_COL
)

stats_after = regression_statistics(
    df_clean,
    HEIGHT_COL,
    BIOMASS_COL
)


print("\n" + "=" * 60)
print("BEFORE HUBER OUTLIER REMOVAL")
print("=" * 60)

print(
    f"R2         : {stats_before['r2']:.3f}"
)

print(
    f"n total    : {stats_before['n']}"
)

print(
    f"n outliers : {n_huber_outliers}"
)

print(
    f"Threshold  : {HUBER_EPSILON}"
)


print("\n" + "=" * 60)
print("AFTER HUBER OUTLIER REMOVAL")
print("=" * 60)

print(
    f"R2      : {stats_after['r2']:.3f}"
)

print(
    f"n clean : {stats_after['n']}"
)

# ============================================================
# CELL 15. FILTERING SUMMARY TABLE
# ============================================================

filter_summary = pd.DataFrame({
    "Filtering step": [
        "Original dataset",
        "Erroneous height <= 0",
        "Missing/non-numeric height",
        "Dataset entering Huber filtering",
        "Huber outliers",
        "Final retained dataset"
    ],
    "n": [
        n_original,
        n_height_le_zero,
        n_height_missing,
        len(df_height_clean),
        n_huber_outliers,
        n_clean
    ]
})

display(filter_summary)


# ============================================================
# CELL 15B. HUBER EXCLUSION AUDIT TABLE
# Date, season, treatment, and biomass range
# ============================================================

# ------------------------------------------------------------
# Identify metadata columns automatically
# Edit candidates here only if your CSV uses another name.
# ------------------------------------------------------------

def find_column(data, candidates):
    for col in candidates:
        if col in data.columns:
            return col
    return None


DATE_COL = find_column(
    df_outliers,
    ["Date", "date", "Harvest_Date", "Harvest Date", "harvest_date"]
)

SEASON_COL = find_column(
    df_outliers,
    ["season", "Season", "SEASON"]
)

PLOT_COL = find_column(
    df_outliers,
    ["Plot", "plot", "Plot_ID", "plot_id"]
)

TREATMENT_COL = find_column(
    df_outliers,
    [
        "Treatment",
        "treatment",
        "Harvest Interval",
        "Harvest_Interval",
        "harvest_interval"
    ]
)

STRIP_COL = find_column(
    df_outliers,
    ["Strip", "strip", "Strip_ID", "strip_id"]
)


# ------------------------------------------------------------
# Working copy
# ------------------------------------------------------------

outlier_audit = df_outliers.copy()


# ------------------------------------------------------------
# Date
# ------------------------------------------------------------

if DATE_COL is not None:
    outlier_audit[DATE_COL] = pd.to_datetime(
        outlier_audit[DATE_COL],
        errors="coerce"
    )


# ------------------------------------------------------------
# Treatment
#
# If no explicit treatment column exists, derive treatment
# from plot identifiers such as:
# 14.1 -> 14
# 28.3 -> 28
# 42.4 -> 42
# ------------------------------------------------------------

if TREATMENT_COL is not None:

    outlier_audit["Treatment_days"] = (
        pd.to_numeric(
            outlier_audit[TREATMENT_COL],
            errors="coerce"
        )
    )

elif PLOT_COL is not None:

    outlier_audit["Treatment_days"] = (
        outlier_audit[PLOT_COL]
        .astype(str)
        .str.extract(
            r"^(\d+)",
            expand=False
        )
        .pipe(
            pd.to_numeric,
            errors="coerce"
        )
    )

else:
    outlier_audit["Treatment_days"] = np.nan


# ------------------------------------------------------------
# Biomass ranges
#
# Fixed ranges are preferable here because they retain the
# same biological interpretation regardless of the number of
# excluded observations.
# ------------------------------------------------------------

biomass_bins = [
    -np.inf,
    1500,
    3000,
    4500,
    6000,
    np.inf
]

biomass_labels = [
    "<1,500",
    "1,500-2,999",
    "3,000-4,499",
    "4,500-5,999",
    ">=6,000"
]

outlier_audit["Biomass_range_kg_DM_ha"] = pd.cut(
    outlier_audit[BIOMASS_COL],
    bins=biomass_bins,
    labels=biomass_labels,
    right=False
)


# ============================================================
# A. CONCISE MANUSCRIPT TABLE
# One row per date × season × treatment
# ============================================================

group_cols = []

if DATE_COL is not None:
    group_cols.append(DATE_COL)

if SEASON_COL is not None:
    group_cols.append(SEASON_COL)

group_cols.append("Treatment_days")


if len(group_cols) > 0:

    exclusion_table = (
        outlier_audit
        .groupby(
            group_cols,
            dropna=False
        )
        .agg(
            n_excluded=(BIOMASS_COL, "size"),
            biomass_min=(BIOMASS_COL, "min"),
            biomass_max=(BIOMASS_COL, "max"),
            biomass_mean=(BIOMASS_COL, "mean"),
            biomass_median=(BIOMASS_COL, "median")
        )
        .reset_index()
    )

    exclusion_table["Biomass range (kg DM ha-1)"] = (
        exclusion_table["biomass_min"]
        .round(0)
        .astype("Int64")
        .astype(str)
        + "–" +
        exclusion_table["biomass_max"]
        .round(0)
        .astype("Int64")
        .astype(str)
    )

    exclusion_table["Mean biomass (kg DM ha-1)"] = (
        exclusion_table["biomass_mean"].round(0)
    )

    exclusion_table["Median biomass (kg DM ha-1)"] = (
        exclusion_table["biomass_median"].round(0)
    )

    exclusion_table = exclusion_table.drop(
        columns=[
            "biomass_min",
            "biomass_max",
            "biomass_mean",
            "biomass_median"
        ]
    )

    rename_map = {
        "Treatment_days": "Treatment (days)",
        "n_excluded": "n excluded"
    }

    if DATE_COL is not None:
        rename_map[DATE_COL] = "Harvest date"

    if SEASON_COL is not None:
        rename_map[SEASON_COL] = "Season"

    exclusion_table = exclusion_table.rename(
        columns=rename_map
    )

    # Format date only
    if "Harvest date" in exclusion_table.columns:
        exclusion_table["Harvest date"] = (
            pd.to_datetime(
                exclusion_table["Harvest date"],
                errors="coerce"
            )
            .dt.strftime("%Y-%m-%d")
        )

    print("\n" + "=" * 80)
    print("HUBER EXCLUSIONS BY HARVEST DATE, SEASON, AND TREATMENT")
    print("=" * 80)

    display(exclusion_table)


# ============================================================
# B. SUMMARY BY SEASON
# ============================================================

if SEASON_COL is not None:

    exclusion_by_season = (
        outlier_audit
        .groupby(
            SEASON_COL,
            dropna=False
        )
        .agg(
            n_excluded=(BIOMASS_COL, "size"),
            biomass_min=(BIOMASS_COL, "min"),
            biomass_max=(BIOMASS_COL, "max")
        )
        .reset_index()
    )

    exclusion_by_season["Biomass range (kg DM ha-1)"] = (
        exclusion_by_season["biomass_min"]
        .round(0)
        .astype("Int64")
        .astype(str)
        + "–" +
        exclusion_by_season["biomass_max"]
        .round(0)
        .astype("Int64")
        .astype(str)
    )

    exclusion_by_season = exclusion_by_season[
        [
            SEASON_COL,
            "n_excluded",
            "Biomass range (kg DM ha-1)"
        ]
    ].rename(
        columns={
            SEASON_COL: "Season",
            "n_excluded": "n excluded"
        }
    )

    print("\nExclusions by season")
    display(exclusion_by_season)


# ============================================================
# C. SUMMARY BY TREATMENT
# ============================================================

exclusion_by_treatment = (
    outlier_audit
    .groupby(
        "Treatment_days",
        dropna=False
    )
    .agg(
        n_excluded=(BIOMASS_COL, "size"),
        biomass_min=(BIOMASS_COL, "min"),
        biomass_max=(BIOMASS_COL, "max")
    )
    .reset_index()
)

exclusion_by_treatment["Biomass range (kg DM ha-1)"] = (
    exclusion_by_treatment["biomass_min"]
    .round(0)
    .astype("Int64")
    .astype(str)
    + "–" +
    exclusion_by_treatment["biomass_max"]
    .round(0)
    .astype("Int64")
    .astype(str)
)

exclusion_by_treatment = exclusion_by_treatment[
    [
        "Treatment_days",
        "n_excluded",
        "Biomass range (kg DM ha-1)"
    ]
].rename(
    columns={
        "Treatment_days": "Treatment (days)",
        "n_excluded": "n excluded"
    }
)

print("\nExclusions by treatment")
display(exclusion_by_treatment)


# ============================================================
# D. SUMMARY BY BIOMASS RANGE
# ============================================================

exclusion_by_biomass_range = (
    outlier_audit
    .groupby(
        "Biomass_range_kg_DM_ha",
        observed=False
    )
    .size()
    .rename("n excluded")
    .reset_index()
    .rename(
        columns={
            "Biomass_range_kg_DM_ha":
                "Biomass range (kg DM ha-1)"
        }
    )
)

print("\nExclusions by biomass range")
display(exclusion_by_biomass_range)


# ============================================================
# E. OPTIONAL ROW-LEVEL AUDIT TABLE
# ============================================================

audit_cols = []

if DATE_COL is not None:
    audit_cols.append(DATE_COL)

if SEASON_COL is not None:
    audit_cols.append(SEASON_COL)

audit_cols.append("Treatment_days")

if PLOT_COL is not None:
    audit_cols.append(PLOT_COL)

if STRIP_COL is not None:
    audit_cols.append(STRIP_COL)

audit_cols.extend(
    [
        HEIGHT_COL,
        BIOMASS_COL,
        "Huber_Predicted_Biomass",
        "Huber_Residual",
        "Huber_Scaled_Residual",
        "Biomass_range_kg_DM_ha"
    ]
)

outlier_audit_detail = (
    outlier_audit[audit_cols]
    .copy()
)

print("\nDetailed excluded-observation audit")
display(outlier_audit_detail)


# ============================================================
# CELL 16. FINAL PUBLICATION-STYLE FIGURE
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5.8),
    sharex=True,
    sharey=True
)


# ============================================================
# DEFINE COMMON AXIS LIMITS
# ============================================================

x_max = df_height_clean[
    HEIGHT_COL
].max()

y_max = df_height_clean[
    BIOMASS_COL
].max()


x_upper = (
    np.ceil(
        x_max / 100
    )
    * 100
)


y_upper = (
    np.ceil(
        y_max / 1000
    )
    * 1000
)


x_grid = np.linspace(
    0,
    x_upper,
    500
).reshape(
    -1,
    1
)


# ============================================================
# PANEL A
# BEFORE HUBER OUTLIER REMOVAL
# ============================================================

ax = axes[0]


# ------------------------------------------------------------
# CLEAN OBSERVATIONS
# ------------------------------------------------------------

obs_handle = ax.scatter(

    df_clean[HEIGHT_COL],

    df_clean[BIOMASS_COL],

    s=12,

    c="black",

    alpha=0.85,

    edgecolors="none",

    label="Observation",

    zorder=2
)


# ------------------------------------------------------------
# HUBER OUTLIERS
# Red cross symbols
# ------------------------------------------------------------

outlier_handle = ax.scatter(

    df_outliers[HEIGHT_COL],

    df_outliers[BIOMASS_COL],

    s=42,

    c="red",

    marker="x",

    linewidths=1.3,

    label="Detected outliers",

    zorder=4
)


# ------------------------------------------------------------
# LINEAR REGRESSION BEFORE FILTERING
# ------------------------------------------------------------

y_reg_before = (
    stats_before["model"]
    .predict(
        x_grid
    )
)


reg_handle, = ax.plot(

    x_grid.ravel(),

    y_reg_before,

    color="black",

    linewidth=1.8,

    label="Huber regression",

    zorder=3
)


# ------------------------------------------------------------
# METRICS BOX
# TOP LEFT
# ------------------------------------------------------------

metrics_a = (

    f"$R^2$ = "
    f"{stats_before['r2']:.2f}\n"

    f"$n$ total = "
    f"{stats_before['n']}\n"

    f"$n$ outliers = "
    f"{n_huber_outliers}\n"

    f"Huber threshold = "
    f"{HUBER_EPSILON:.1f}"
)


ax.text(

    0.035,
    0.95,

    metrics_a,

    transform=ax.transAxes,

    ha="left",

    va="top",

    fontsize=9,

    bbox=dict(

        boxstyle="square,pad=0.55",

        facecolor="white",

        edgecolor="0.3",

        linewidth=0.8
    )
)


# ------------------------------------------------------------
# LEGEND
# BOTTOM RIGHT
# WITH BOX
# ------------------------------------------------------------

legend_a = ax.legend(

    handles=[
        obs_handle,
        outlier_handle,
        reg_handle
    ],

    labels=[
        "Observation",
        "Detected outliers",
        "Huber regression"
    ],

    loc="lower right",

    frameon=True,

    fancybox=False,

    framealpha=1.0,

    facecolor="white",

    edgecolor="0.3",

    handlelength=2.4,

    borderpad=0.8,

    borderaxespad=0.8
)


legend_a.get_frame().set_linewidth(
    0.8
)


# ------------------------------------------------------------
# PANEL TITLE AND AXES
# ------------------------------------------------------------

ax.set_title(
    "(a) Before outlier removal",
    fontweight="bold"
)

ax.set_xlabel(
    "Plant Height (mm)"
)

ax.set_ylabel(
    r"Total Biomass (kg ha$^{-1}$)"
)


# ============================================================
# PANEL B
# AFTER HUBER OUTLIER REMOVAL
# ============================================================

ax = axes[1]


# ------------------------------------------------------------
# CLEAN OBSERVATIONS
# ------------------------------------------------------------

obs_handle_b = ax.scatter(

    df_clean[HEIGHT_COL],

    df_clean[BIOMASS_COL],

    s=12,

    c="black",

    alpha=0.85,

    edgecolors="none",

    label="Observation",

    zorder=2
)


# ------------------------------------------------------------
# LINEAR REGRESSION AFTER FILTERING
# ------------------------------------------------------------

y_reg_after = (
    stats_after["model"]
    .predict(
        x_grid
    )
)


reg_handle_b, = ax.plot(

    x_grid.ravel(),

    y_reg_after,

    color="black",

    linewidth=1.8,

    label="Huber regression",

    zorder=3
)


# ------------------------------------------------------------
# METRICS BOX
# TOP LEFT
# ------------------------------------------------------------

metrics_b = (

    f"$R^2$ = "
    f"{stats_after['r2']:.2f}\n"

    f"$n$ clean = "
    f"{stats_after['n']}"
)


ax.text(

    0.035,
    0.95,

    metrics_b,

    transform=ax.transAxes,

    ha="left",

    va="top",

    fontsize=9,

    bbox=dict(

        boxstyle="square,pad=0.55",

        facecolor="white",

        edgecolor="0.3",

        linewidth=0.8
    )
)


# ------------------------------------------------------------
# LEGEND
# BOTTOM RIGHT
# WITH BOX
# ------------------------------------------------------------

legend_b = ax.legend(

    handles=[
        obs_handle_b,
        reg_handle_b
    ],

    labels=[
        "Observation",
        "Huber regression"
    ],

    loc="lower right",

    frameon=True,

    fancybox=False,

    framealpha=1.0,

    facecolor="white",

    edgecolor="0.3",

    handlelength=2.4,

    borderpad=0.8,

    borderaxespad=0.8
)


legend_b.get_frame().set_linewidth(
    0.8
)


# ------------------------------------------------------------
# PANEL TITLE AND AXES
# ------------------------------------------------------------

ax.set_title(
    "(b) After outlier removal",
    fontweight="bold"
)

ax.set_xlabel(
    "Plant Height (mm)"
)

ax.set_ylabel(
    r"Total Biomass (kg ha$^{-1}$)"
)


# ============================================================
# CELL 17. COMMON FIGURE FORMATTING
# ============================================================

for ax in axes:

    ax.set_xlim(
        0,
        x_upper
    )

    ax.set_ylim(
        0,
        y_upper
    )

    ax.tick_params(
        direction="out",
        width=0.8
    )

    for spine in ax.spines.values():

        spine.set_linewidth(
            0.8
        )

    ax.grid(
        False
    )


plt.tight_layout(
    w_pad=3
)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

from pathlib import Path

OUTPUT_DIR = Path(
    r"C:\Users\ag2c2\OneDrive - University of Missouri"
    r"\RESEARCH\Testbed Paper\data\Huber"
)

# Create folder automatically if it does not exist
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"All outputs will be saved to:\n{OUTPUT_DIR}")
# ============================================================
# CELL 18. SAVE FINAL JPEG
# ============================================================

output_figure = (
    OUTPUT_DIR / "Huber_Filtering_Before_After.jpg"
)

plt.savefig(
    output_figure,
    dpi=600,
    format="jpg",
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print(
    f"Figure saved as:\n{output_figure}"
)

# ============================================================
# CELL 19. EXPORT FILTERED DATASETS AND EXCLUSION TABLES
# ============================================================

df_clean.to_csv(
    OUTPUT_DIR / "TB_24_25_VI_Clean.csv",
    index=False
)

df_outliers.to_csv(
    OUTPUT_DIR / "TB_24_25_VI_Huber_Outliers.csv",
    index=False
)

erroneous_height_df.to_csv(
    OUTPUT_DIR / "TB_24_25_VI_Erroneous_Heights.csv",
    index=False
)

filter_summary.to_csv(
    OUTPUT_DIR / "TB_24_25_VI_Filtering_Summary.csv",
    index=False
)


# ------------------------------------------------------------
# Export exclusion-audit tables
# ------------------------------------------------------------

if "exclusion_table" in globals():

    exclusion_table.to_csv(
        OUTPUT_DIR / "Huber_Exclusions_By_Date_Season_Treatment.csv",
        index=False,
        encoding="utf-8-sig"
    )


if "exclusion_by_season" in globals():

    exclusion_by_season.to_csv(
        OUTPUT_DIR / "Huber_Exclusions_By_Season.csv",
        index=False,
        encoding="utf-8-sig"
    )


exclusion_by_treatment.to_csv(
    OUTPUT_DIR / "Huber_Exclusions_By_Treatment.csv",
    index=False,
    encoding="utf-8-sig"
)


exclusion_by_biomass_range.to_csv(
    OUTPUT_DIR / "Huber_Exclusions_By_Biomass_Range.csv",
    index=False,
    encoding="utf-8-sig"
)


outlier_audit_detail.to_csv(
    OUTPUT_DIR / "Huber_Excluded_Observations_Detail.csv",
    index=False,
    encoding="utf-8-sig"
)


print("=" * 70)
print("All Huber outputs exported successfully.")
print("=" * 70)
print(f"Output directory:\n{OUTPUT_DIR}")


# ------------------------------------------------------------
# Export exclusion-audit tables
# ------------------------------------------------------------

if "exclusion_table" in globals():

    exclusion_table.to_csv(
        "Huber_Exclusions_By_Date_Season_Treatment.csv",
        index=False,
        encoding="utf-8-sig"
    )


if "exclusion_by_season" in globals():

    exclusion_by_season.to_csv(
        "Huber_Exclusions_By_Season.csv",
        index=False,
        encoding="utf-8-sig"
    )


exclusion_by_treatment.to_csv(
    "Huber_Exclusions_By_Treatment.csv",
    index=False,
    encoding="utf-8-sig"
)


exclusion_by_biomass_range.to_csv(
    "Huber_Exclusions_By_Biomass_Range.csv",
    index=False,
    encoding="utf-8-sig"
)


outlier_audit_detail.to_csv(
    "Huber_Excluded_Observations_Detail.csv",
    index=False,
    encoding="utf-8-sig"
)


print(
    "Filtered datasets and exclusion-audit tables exported."
)


# ============================================================
# CELL 20. FINAL REPORT
# ============================================================

print(
    "\n" + "=" * 65
)

print(
    "FINAL DATA FILTERING REPORT"
)

print(
    "=" * 65
)


print(
    f"Original observations                : "
    f"{n_original:,}"
)

print(
    f"Erroneous heights <= 0 removed       : "
    f"{n_height_le_zero:,}"
)

print(
    f"Missing/non-numeric heights excluded : "
    f"{n_height_missing:,}"
)

print(
    f"Records entering Huber filtering     : "
    f"{len(df_height_clean):,}"
)

print(
    f"Huber threshold                      : "
    f"{HUBER_EPSILON:.2f}"
)

print(
    f"Huber outliers removed               : "
    f"{n_huber_outliers:,}"
)

print(
    f"Final clean observations             : "
    f"{n_clean:,}"
)

print(
    f"R2 before filtering                  : "
    f"{stats_before['r2']:.3f}"
)

print(
    f"R2 after filtering                   : "
    f"{stats_after['r2']:.3f}"
)

print(
    "=" * 65
)
```
